```python
import websockets

async def listen():
    # Use wss:// (secure WebSocket) to match ngrok's HTTPS requirement
    uri = 'wss://e130-152-56-156-52.ngrok-free.app/room/ws'
    
    # Headers to bypass ngrok's interstitial browser landing page
    extra_headers = {
        "ngrok-skip-browser-warning": "true",
        "User-Agent": "Mozilla/5.0"
    }
    
    async with websockets.connect(uri, additional_headers=extra_headers) as ws:
        # 1. Send STOMP CONNECT frame
        await ws.send("CONNECT\naccept-version:1.2\nheart-beat:0,0\n\n\x00")
        conn_ack = await ws.recv()
        print("Connected to STOMP Broker:\n", conn_ack)

        # 2. Send STOMP SUBSCRIBE frame
        await ws.send("SUBSCRIBE\nid:sub-0\ndestination:/topic/risk\n\n\x00")
        print("Subscribed to /topic/risk! Listening for periodic broadcasts...\n")

        # 3. Stream incoming messages
        while True:
            msg = await ws.recv()
            print("RECEIVED BROADCAST:\n", msg)

await listen()
```
